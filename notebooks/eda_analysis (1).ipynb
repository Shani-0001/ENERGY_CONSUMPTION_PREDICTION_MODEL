1️. What is the distribution of Energy Consumption?

```python
plt.figure(figsize=(8,5))
sns.histplot(df['EnergyConsumption'], kde=True)
plt.title("Distribution of Energy Consumption")
plt.show()
```

Interpretation :

Shows how energy consumption values are spread.

If right-skewed → high consumption occurs less frequently.

Helps detect outliers and data distribution.

2.How does Temperature affect Energy Consumption?

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='Temperature', y='EnergyConsumption', data=df)
plt.title("Temperature vs Energy Consumption")
plt.show()
```

Interpretation:

Shows relationship between temperature and energy usage.

Higher temperature may increase HVAC usage and energy consumption.

3.How does Humidity influence Energy Consumption?

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='Humidity', y='EnergyConsumption', data=df)
plt.title("Humidity vs Energy Consumption")
plt.show()
```

Interpretation:

Higher humidity may increase cooling demand.

Can identify correlation between humidity and energy usage.

4.Does Occupancy impact Energy Consumption?

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='Occupancy', y='EnergyConsumption', data=df)
plt.title("Occupancy vs Energy Consumption")
plt.show()
```

Interpretation:

More occupants usually increase energy use.

Boxplot shows median, spread, and outliers.

5.How does HVAC Usage affect Energy Consumption?

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='HVACUsage', y='EnergyConsumption', data=df)
plt.title("HVAC Usage vs Energy Consumption")
plt.show()
```

Interpretation:

HVAC systems are major energy consumers.

Strong positive relationship expected.

6.How does Lighting Usage affect Energy Consumption?

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='LightingUsage', y='EnergyConsumption', data=df)
plt.title("Lighting Usage vs Energy Consumption")
plt.show()
```

Interpretation:

Higher lighting usage contributes to total energy usage.

Relationship may be weaker than HVAC.

7.How does Renewable Energy affect total consumption?

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='RenewableEnergy', y='EnergyConsumption', data=df)
plt.title("Renewable Energy vs Energy Consumption")
plt.show()
```

Interpretation:

Renewable sources may reduce grid energy consumption.

Helps identify sustainability impact.

8.What is the correlation between all numerical features?

```python
plt.figure(figsize=(12,7))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

Interpretation:

Shows relationship strength between variables.

Helps select features for machine learning.

9.How does Energy Consumption change by Hour of the Day?

```python
plt.figure(figsize=(10,5))
df.groupby("Timestamp")["EnergyConsumption"].mean().plot()
plt.title("Energy Consumption by Hour")
plt.xlabel("Timestamp")
plt.ylabel("Average Energy Consumption")
plt.show()
```

Interpretation:

Identifies peak energy usage hours.

Usually higher during working hours.

10.How does Energy Consumption vary across Months?

```python
plt.figure(figsize=(10,5))
df.groupby("DayOfWeek")["EnergyConsumption"].mean().plot()
plt.title("Energy Consumption by Month")
plt.xlabel("DayOfWeek")
plt.ylabel("Average Energy Consumption")
plt.show()
```

Interpretation:

Shows seasonal patterns.

Higher consumption in summer/winter due to HVAC usage.
